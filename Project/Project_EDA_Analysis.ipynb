```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('data.csv')
```

```python
print("Dataset Shape:", df.shape)
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df = df.drop(columns=['Unnamed: 32'])
df = df.drop(columns=['id'])
```

```python
print("Dataset Shape:", df.shape)
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
```

```python
print(df['diagnosis'].value_counts())
print(df['diagnosis'].value_counts(normalize=True) * 100)
```

```python
plt.figure(figsize=(6, 6))

df['diagnosis'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Distribution of Benign and Malignant Cases')
plt.ylabel('')
plt.show()
```

```python
features = df.drop(columns=['diagnosis'])

features.describe().T
```

```python
features.hist(
    figsize=(20, 20),
    bins=20
)

plt.suptitle('Distribution of 30 Numerical Features', fontsize=16)
plt.show()
```

```python
plt.figure(figsize=(20, 10))

sns.boxplot(data=features)

plt.xticks(rotation=90)
plt.title('Boxplot of 30 Numerical Features')
plt.show()
```

```python
plt.figure(figsize=(18, 14))

correlation = features.corr()

sns.heatmap(
    correlation,
    cmap='coolwarm',
    center=0
)

plt.title('Correlation Heatmap of 30 Features')
plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.boxplot(
    x='diagnosis',
    y='radius_mean',
    data=df
)

plt.title('Radius Mean by Diagnosis')
plt.xlabel('Diagnosis')
plt.ylabel('Mean Radius')
plt.show()
```

```python
df['diagnosis_numeric'] = df['diagnosis'].map({
    'B': 0,
    'M': 1
})
```

```python
target_correlation = df.corr(numeric_only=True)['diagnosis_numeric'].drop('diagnosis_numeric')

target_correlation = target_correlation.abs().sort_values(ascending=False)

print(target_correlation)
```

```python
top_10 = target_correlation.head(10)

plt.figure(figsize=(10, 6))

top_10.sort_values().plot(kind='barh')

plt.title('Top 10 Features Correlated with Diagnosis')
plt.xlabel('Absolute Correlation')
plt.ylabel('Feature')
plt.show()
```
