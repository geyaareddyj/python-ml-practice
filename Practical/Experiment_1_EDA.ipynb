```python
import pandas as pd
from IPython.display import display

#File name

csv_path= "placement_predict_50k_Dataset.csv"

df= pd.read_csv(csv_path)
```

```python
# Display first 5 rows
print("First 5 Rows")
display(df.head())
```

```python
print("\n Dataset Shape")

print(df.shape)
```

```python
print("\nColumn Names")

print(df.columns)
```

```python
print("\n Data types")

display(df.dtypes)
```

```python
print("\n Dataset Information")

display(df.info())
```

```python
print("\nStatistical Summary")

display(df.describe())
```

```python
print("\nMissing Values")

display(df.isnull().sum())
```

```python
print("\nDuplicate Rows")

print(df.duplicated().sum())
```

```python
print("\nPlacement Status Distribution")

placement_status_counts = df["PlacementStatus"].value_counts().reset_index()
placement_status_counts.columns = ['PlacementStatus', 'Count']
display(placement_status_counts)
```

```python
print("\nPlacement Status Percentage")

placement_status_per=(df["PlacementStatus"].value_counts(normalize=True)*100).reset_index()
placement_status_per.columns = ['PlacementStatus', 'Percentage']
display(placement_status_per)
```
